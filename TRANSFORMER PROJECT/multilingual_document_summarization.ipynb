```python
import torch
from transformers import pipeline
from datasets import load_dataset
```

```python
# Load only the first 100 examples from the training set
dataset = load_dataset("mlsum", "fr", split="train[:100]")

# Inspect the dataset
print(dataset)
```

```python
# Load the multilingual summarization pipeline
summarizer = pipeline("summarization", model="google/mt5-small")
```

```python
# Summarize the first 5 articles
for i in range(5):
    article = dataset[i]["text"]
    reference_summary = dataset[i]["summary"]

    # Generate a summary using the model
    generated_summary = summarizer(article, max_length=50, min_length=25, do_sample=False)[0]['summary_text']

    # Print results
    print(f"Article {i+1}:")
    print("Reference Summary:", reference_summary)
    print("Generated Summary:", generated_summary)
    print("-" * 80)
```

```python
# Load the translation pipeline
translator = pipeline("translation", model="facebook/nllb-200-distilled-600M")

# Translate the generated summary to English
translated_summary = translator(generated_summary, src_lang="fra_Latn", tgt_lang="eng_Latn")[0]['translation_text']
print("Translated Summary in English:", translated_summary)
```
