```python
from transformers import MarianMTModel, MarianTokenizer

# MarianMT provides pre-trained models for many language pairs (e.g., English-to-French, French-to-English, etc.).
# The MarianTokenizer is responsible for converting raw text into tokens that the MarianMTModel can process.
# It also converts the model’s output tokens back into human-readable text.
```

```python
# Step 1: Choose a pre-trained model and tokenizer
model_name = "Helsinki-NLP/opus-mt-en-fr"
```

```python
# Load the tokenizer and model
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)
```

```python
# Step 2: Define a function to translate English to French
def translate_english_to_french(text):
    # Tokenize the input text
    tokenized_text = tokenizer.prepare_seq2seq_batch([text], return_tensors="pt") # pt = pytorch
    # The prepare_seq2seq_batch method prepares the input text for a sequence-to-sequence model.

    # Generate translation using the model
    translated_tokens = model.generate(**tokenized_text)

    # Decode the translated tokens into text
    translated_text = tokenizer.decode(translated_tokens[0], skip_special_tokens=True)
    # Converts the model’s output tokens back into human-readable text.

    return translated_text

# 1) Tokenization: Converts text into numerical tokens for the model.

# 2) Model Inference: Generates translated tokens using the MarianMT model.

# 3) Decoding: Converts the model’s output tokens back into readable text.

# 4) Special Tokens: Skipped in the final output for cleaner results.
```

```python
# Step 3: Test the translation function
if __name__ == "__main__":
    # Input English text
    english_text = input("Enter English text to translate: ")

    # Translate to French
    french_translation = translate_english_to_french(english_text)

    # Print the results
    print(f"English: {english_text}")
    print(f"French: {french_translation}")
```
